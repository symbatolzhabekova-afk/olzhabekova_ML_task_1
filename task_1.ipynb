```python
import pandas as pd

YOUR_ID = 2650

raw = pd.read_csv("data/almaty_apartments_raw (3).csv")

ids = raw["listing_id"].drop_duplicates().sample(
    frac=0.9,
    random_state=YOUR_ID
)

df = raw[raw["listing_id"].isin(ids)].reset_index(drop=True)

print(df.shape)
print(df.dtypes)
print(df.isna().sum())
print(df.describe(include="all").T)
```

```python

df_clean = df.copy()

df_clean["listed_date"] = pd.to_datetime(
    df_clean["listed_date"],
    errors="coerce"
)

df_clean["rooms"] = (
    df_clean["rooms"]
    .replace({"studio": "1"})
    .astype(int)
)

df_clean["area_m2"] = (
    df_clean["area_m2"]
    .str.replace(",", ".", regex=False)
    .str.replace("m²", "", regex=False)
    .str.strip()
)

df_clean["area_m2"] = pd.to_numeric(
    df_clean["area_m2"],
    errors="coerce"
)

df_clean["price_kzt"] = (
    df_clean["price_kzt"]
    .str.replace(" ", "", regex=False)
    .str.replace("₸", "", regex=False)
    .str.replace(",", "", regex=False)
    .str.strip()
)

df_clean["price_kzt"] = pd.to_numeric(
    df_clean["price_kzt"],
    errors="coerce"
)

print(df_clean.dtypes)
```

```python
from sklearn.model_selection import train_test_split

train_df, test_df = train_test_split(
    df_clean,
    test_size=0.20,
    random_state=2650
)

print("Train shape:", train_df.shape)
print("Test shape:", test_df.shape)
```

```python
# Calculate medians using TRAIN data only
year_built_median = train_df["year_built"].median()
ceiling_height_median = train_df["ceiling_height_m"].median()

print("Train median year_built:", year_built_median)
print("Train median ceiling_height_m:", ceiling_height_median)

# Fill missing values in both datasets
train_df["year_built"] = train_df["year_built"].fillna(year_built_median)
test_df["year_built"] = test_df["year_built"].fillna(year_built_median)

train_df["ceiling_height_m"] = train_df["ceiling_height_m"].fillna(
    ceiling_height_median
)
test_df["ceiling_height_m"] = test_df["ceiling_height_m"].fillna(
    ceiling_height_median
)
```

```python
print("Missing values in TRAIN:")
print(train_df.isna().sum())

print("\nMissing values in TEST:")
print(test_df.isna().sum())
```

```python
def iqr_bounds(series):
    Q1 = series.quantile(0.25)
    Q3 = series.quantile(0.75)
    IQR = Q3 - Q1
    
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    
    return lower, upper


lower_area, upper_area = iqr_bounds(train_df["area_m2"])

print("Area lower bound:", lower_area)
print("Area upper bound:", upper_area)

print("\nPotential area outliers:")
print(
    train_df.loc[
        (train_df["area_m2"] < lower_area) |
        (train_df["area_m2"] > upper_area),
        ["listing_id", "area_m2", "rooms", "price_kzt"]
    ].sort_values("area_m2", ascending=False)
)
```

```python
# IQR bounds calculated from TRAIN data
Q1_area = train_df["area_m2"].quantile(0.25)
Q3_area = train_df["area_m2"].quantile(0.75)

IQR_area = Q3_area - Q1_area

lower_area = max(0, Q1_area - 1.5 * IQR_area)
upper_area = Q3_area + 1.5 * IQR_area

print("Q1:", Q1_area)
print("Q3:", Q3_area)
print("IQR:", IQR_area)
print("Lower bound:", lower_area)
print("Upper bound:", upper_area)
```

```python
train_df["area_m2"] = train_df["area_m2"].clip(
    lower=lower_area,
    upper=upper_area
)

test_df["area_m2"] = test_df["area_m2"].clip(
    lower=lower_area,
    upper=upper_area
)

print("Maximum area in train:", train_df["area_m2"].max())
print("Maximum area in test:", test_df["area_m2"].max())
```

```python
# Building age at the time of listing
train_df["age_at_listing"] = (
    train_df["listed_date"].dt.year - train_df["year_built"]
)

test_df["age_at_listing"] = (
    test_df["listed_date"].dt.year - test_df["year_built"]
)

# Price per square meter
train_df["price_per_m2"] = (
    train_df["price_kzt"] / train_df["area_m2"]
)

test_df["price_per_m2"] = (
    test_df["price_kzt"] / test_df["area_m2"]
)

print(train_df[[
    "listed_date",
    "year_built",
    "age_at_listing",
    "area_m2",
    "price_kzt",
    "price_per_m2"
]].head())
```

```python
from sklearn.preprocessing import OneHotEncoder

categorical_features = [
    "district",
    "building_type"
]

encoder = OneHotEncoder(
    handle_unknown="ignore",
    sparse_output=False
)

train_encoded = encoder.fit_transform(
    train_df[categorical_features]
)

test_encoded = encoder.transform(
    test_df[categorical_features]
)

print("Train encoded shape:", train_encoded.shape)
print("Test encoded shape:", test_encoded.shape)
print("Encoded features:", encoder.get_feature_names_out())
```

```python
district_mapping = {
    "alatau": "Alatau",
    "Алатауский": "Alatau",
    "Alatauskiy": "Alatau",

    "almaly": "Almaly",
    "Алмалинский": "Almaly",
    "Almalinskiy": "Almaly",

    "auezov": "Auezov",
    "Ауэзовский": "Auezov",
    "Auezovskiy": "Auezov",

    "bostandyk": "Bostandyk",
    "Бостандыкский": "Bostandyk",
    "Bostandykskiy": "Bostandyk",

    "medeu": "Medeu",
    "медеуский": "Medeu",
    "Медеуский": "Medeu",
    "Medeuskiy": "Medeu",

    "nauryzbay": "Nauryzbay",
    "Наурызбайский": "Nauryzbay",
    "Nauryzbayskiy": "Nauryzbay",

    "turksib": "Turksib",
    "Турксибский": "Turksib",
    "Turksibskiy": "Turksib",

    "zhetysu": "Zhetysu",
    "Жетысуский": "Zhetysu",
    "Zhetysuskiy": "Zhetysu"
}

def normalize_district(value):
    value = str(value).strip()
    
    # First handle case-insensitive Latin variants
    key = value.lower()
    
    if key in district_mapping:
        return district_mapping[key]
    
    # Handle Cyrillic variants
    if value in district_mapping:
        return district_mapping[value]
    
    # Standard capitalization for already-clean values
    return value.capitalize()


train_df["district"] = train_df["district"].apply(normalize_district)
test_df["district"] = test_df["district"].apply(normalize_district)

print("Train districts:")
print(sorted(train_df["district"].unique()))

print("\nTest districts:")
print(sorted(test_df["district"].unique()))
```

```python
from sklearn.preprocessing import OneHotEncoder

categorical_features = [
    "district",
    "building_type"
]

encoder = OneHotEncoder(
    handle_unknown="ignore",
    sparse_output=False
)

train_encoded = encoder.fit_transform(
    train_df[categorical_features]
)

test_encoded = encoder.transform(
    test_df[categorical_features]
)

encoded_feature_names = encoder.get_feature_names_out(
    categorical_features
)

print("Train encoded shape:", train_encoded.shape)
print("Test encoded shape:", test_encoded.shape)
print("\nEncoded features:")
print(encoded_feature_names)
```

```python
numeric_features = [
    "rooms",
    "area_m2",
    "floor",
    "total_floors",
    "year_built",
    "ceiling_height_m",
    "age_at_listing",
    "price_per_m2"
]

X_train_numeric = train_df[numeric_features]
X_test_numeric = test_df[numeric_features]

X_train = pd.concat(
    [
        X_train_numeric.reset_index(drop=True),
        pd.DataFrame(
            train_encoded,
            columns=encoded_feature_names
        )
    ],
    axis=1
)

X_test = pd.concat(
    [
        X_test_numeric.reset_index(drop=True),
        pd.DataFrame(
            test_encoded,
            columns=encoded_feature_names
        )
    ],
    axis=1
)

y_train = train_df["price_kzt"].reset_index(drop=True)
y_test = test_df["price_kzt"].reset_index(drop=True)

print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = X_train.copy()
X_test_scaled = X_test.copy()

# Fit only on TRAIN
X_train_scaled[numeric_features] = scaler.fit_transform(
    X_train[numeric_features]
)

# Apply the same scaler to TEST
X_test_scaled[numeric_features] = scaler.transform(
    X_test[numeric_features]
)

print("X_train_scaled shape:", X_train_scaled.shape)
print("X_test_scaled shape:", X_test_scaled.shape)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LinearRegression

model_pipeline = Pipeline([
    ("model", LinearRegression())
])

model_pipeline.fit(X_train_scaled, y_train)

y_pred = model_pipeline.predict(X_test_scaled)

print("Pipeline fitted successfully.")
print("Number of predictions:", len(y_pred))
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)

print("MAE:", mae)
print("RMSE:", rmse)
print("R²:", r2)
```

# ML Data Preparation and Regression

## 1. Setup and Data Loading

The dataset is loaded and inspected before preprocessing.

## 2. Data Cleaning

The dataset was cleaned by converting dates and numeric columns to appropriate data types. 
Categorical and numeric inconsistencies were handled before model training.

## 3. Train/Test Split

The cleaned dataset was split into training and test sets using an 80/20 ratio with random_state=2650.

## 4. Missing Value Imputation

Missing values in `year_built` and `ceiling_height_m` were replaced using medians calculated from the training data only.

## 5. Outlier Handling

The IQR method was used to identify extreme values in `area_m2`.

Q1 = 45.0 m²  
Q3 = 79.4 m²  
IQR = 34.4 m²  
Upper bound = 131.0 m²

Values above the upper bound were capped at 131.0 m² rather than removing complete observations.

## 6. Feature Engineering

Two additional features were created:

- `age_at_listing` — building age at the time of listing.
- `price_per_m2` — apartment price per square meter.

## 7. Categorical Encoding

The `district` values were normalized to eliminate differences in capitalization, spelling, and language.

One-Hot Encoding was then applied to `district` and `building_type`.

The resulting encoded feature matrix contains 11 categorical features.

## 8. Feature Scaling

StandardScaler was fitted using the training data only and then applied to both training and test sets.

## 9. Model Training

A Linear Regression model was trained using the prepared training data through a Pipeline.

## 10. Model Evaluation

The model was evaluated on the held-out test set using MAE, RMSE, and R².

## 11. Conclusion

The dataset was successfully cleaned and prepared for machine learning.

The final preprocessing included missing-value imputation, IQR-based outlier capping, feature engineering, categorical encoding, and feature scaling.

The final model was trained using 925 training observations and evaluated on 232 test observations.
